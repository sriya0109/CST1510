```python
class student:
    def __init__(self,name,program,mark):
        self.name = name
        self.program = program
        self.mark = mark

students=[]
with open("studentdata.txt","r") as file:
    for line in file:
        name,program, mark= line.strip().split(",")
        s=student(name, program, int(mark))
        students.append(s)
```
